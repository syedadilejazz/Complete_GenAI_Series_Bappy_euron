```python
from llama_index import VectorStoreindex, SimpleDirectoryReader,ServiceContext
from llama_index.llms import HuggingFaceLLM
from llama_index.prompts.prompts import SimpleInputPrompt
#VectorStoreIndex-LlamaIndex creates an in-memory vector database. langchain have same kind of function InMemoryVectorStore
```

```python
!mkdir Data
```

```python
documents=SimpleInputPrompt("/content?Data").load_data()
documents
```

```python
system_prompt="""You are a Q&A assistant . Your goal is to answer question as accuratly as possible based on the instructions and context provided"""
#This will wrap the default prompts that are internal to llama-index
query_wrapper_prompt=SimpleInputPrompt("<|USER|>{query_str}<|ASSISTANT|")
```

```python
!huggingface-cli login
```

```python
import torch

llm=HuggingFaceLLM(
    context_window=4096,#Input size of mistral model
    max_new_tokens=256,
    generate_kwargs={"temperature":0.0,"do_sample":False},
    system_prompt=system_prompt,
    query_wrapper_prompt=query_wrapper_prompt,
    tokenizer_name="mistralai/Mistral-7B-Instruct-v0.1",
    model_name="mistralai/Mistral-7B-Instruct-v0.1",
    device_map="auto",
    #uncomment this if using CUDA to reduce memory usage
    tokenizer_kwargs={"torch_dtype":torch.float16,"load_in_8bit":True})
```

```python
from langchain.embeddings.huggingface import HuggingFaceEmbeddings
from llama_index import ServiceContext
from llama_index.embeddings import LangchainEmbedding

embed_model=LangchainEmbedding(HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2"))
```

```python
service_context=ServiceContext.from_defaults(
    llm=llm,
    embed_model=embed_model,
    chunk_size=1024
)
```

```python
index=VectorStoreIndex.from_documents(documents,service_context=service_context)
```
