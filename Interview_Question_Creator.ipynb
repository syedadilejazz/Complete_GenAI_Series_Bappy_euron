#Prompt Template

```python
prompt_templates="""
You are an expert at creating question based in coding materials and documentation.
Your goal is to prepare a coder or programmer for their exam and coding tests.
you do this by asking questions about the text below:
---------------------
{text}
--------------------

Create questions that will prepare the coders or programeers for their tests.
Make sure not to lose any important information.

QUESTIONS:
"""
```

```python
from langchain.prompts import prompt_template
```

```python
prompt_question=prompt_template(template=prompt_templates,input_variables=["text"])
prompt_question
```

```python
ques_gen_chain=load_summarize_chain(
    llm=llm_ques_gen_pipeline,
    chain_type="refine",
    verbose=True,
    question_prompt=prompt_question,
    refine_prompt=refine_prompt_question
)
```
