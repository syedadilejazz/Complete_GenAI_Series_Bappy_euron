```python
from langchain_huggingface.llms import HuggingFacePipeline
import torch
from transformers import AutoTokenizer, AutoModelForCausalLM, pipeline, BitsAndBytesConfig
from langchain_core.prompts import PromptTemplate
```

```python
model="tiiuae/falcon-7b-instruct"

tokenizer=AutoTokenizer.from_pretrained(model)

pipeline=pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    dtypes=torch.bfloat16,
    truct_remote_code=True,
    device_map="auto",
    max_lenghth=200,
    do_sample=True,
    top_k=10,
    num_return_sequences=1,
    eos_token_id=tokenizer.eos_token_id
)
```

```python
llm=HuggingFacePipeline(pipeline=pipeline,model_kwargs={'temperature':0.5})
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
template="""
You are an intelligent chatbot.Help the folowing question with brillinat answer.
Question:{question}
Answer:"""
```

```python
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser

template = PromptTemplate(
    input_variables=["question"],
    template="""
You are a helpful assistant.

Give me a good shop name for:
{question}
"""
)

chain = template | llm | StrOutputParser()

response = chain.invoke({
    "question": "Give me a good shop name for colorful socks"
})

print(response)
```
