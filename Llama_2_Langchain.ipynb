#Load the LLama 2 Model

We are using Llama 2 Chat Models with 7 Billion Parameters
The basic building block of Langchain is a Large Language Model which takes text as inpout and generates more text

Suppose we want to generate a company name based on the compnay description. In the case, since we eant the output to be more random, we will intialize our model with high temperature.

The Temperature paremeter adjusts the randomness of the output. Hgher values like 0.7 will make the output more random, while lower values like 0.2 will make it more focussed and deterministic

Temperature
0-> Temperature it means model is very safe it is not taking any bets
1-> it will risk it might generate wrong output but it is very creative

```python
model="meta-llama/Llama-2-7b-chat-hf"#private/official repo
#model="daryl149/llama-2-7b-chat-hf"#public repo
```

```python
tokenizer=AutoTokenizer.from_pretrained(model)
```

#Issue:https://github.com/facebookresarch/llama/issues/374

```python
pipeline=transformers.pipeline(
    "text-generation",
    model=model,
    tokenizer=tokenizer,
    dtype=torch.float16,
    trust_remote_code=True,
    device_map="auto",
    max_length=1000,
    do_sample=True,
    top_k=10,
    num_return_sequences=1,
    eos_token_id=tokenizer.eos_token_id
)
```

```python
llm=HuggingFacePipeline(pipeline=pipeline,model_kwargs={'temperature':0})
```

```python
prompt='What would be a good name for a company that makes colorful socks'#LLM will use default promptTemplate here
```

```python
print(llm.invoke(prompt))
```

```python
prompt='What would be a good name for a company that makes colorful socks'#LLM will use default promptTemplate here
print(llm.invoke(prompt))
```

#Prompt Templates

Currentlt in the above applications we are writing an entire prompt, if you are creating a user directed aplication then this is not an ideal case .
Langchain facilities prompt management and optimization

Normalyy when you use a LLM in an aplication, you are not sending user input directlt to the LLM. Instead, you need to take the user input and construct a prompt, and only then send that to the LLM.

In many Large Language Model Application we donot pass the user inout directly to the Large language Model, we add the user input to a large piece of text called prompt template

#Import All the required Libaries

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt = PromptTemplate(
    input_variables=["product"],
    template="What is a good name for a company that makes {product}?"
)

chain = prompt | llm

response = chain.invoke({
    "product": "colorful socks"
})

print(response)
```

```python
prompt = PromptTemplate(
    input_variables=["cusine"],
    template="What would be a good restuarent name for  {cusine}?"
)

chain = prompt | llm

response = chain.invoke({
    "cusine": "Indian"
})

print(response)
```

```python
prompt = PromptTemplate(
    input_variables=["book_name"],
    template="Give me summary of {book_name}?"
)

chain = prompt | llm

response = chain.invoke({
    "book_name": "Alchemist"
})

print(response)
```
