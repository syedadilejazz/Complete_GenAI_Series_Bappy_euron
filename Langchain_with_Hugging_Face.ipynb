```python
model_id=AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-large",device_map="auto")
```

```python
text_pipeline = pipeline(
    "text-generation",
    model=model_id,
    tokenizer=tokenizer,
    max_new_tokens=100,
    max_length=None,           # আগের ডিফল্ট max_length মুছে ফেলার জন্য ✅
    return_full_text=True,
    do_sample=True,
    temperature=0.7
)
```

```python
local_llm=HuggingFacePipeline(pipeline=text_pipeline)
```

```python
prompt=PromptTemplate(
    input_variables=['product'],
    template="What is a good name for a company that makes {product}?")
```

```python
chain = prompt | local_llm
repose=chain.invoke({"product":"colorful socks"})
print("Response",repose)
```
